```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Set random seed for reproducibility
np.random.seed(42)

def generate_lhc_event_data(num_events=200):
    """
    Generates synthetic LHC collision feature vectors:
    Features: [Transverse Momentum (pT), Pseudorapidity (eta)]
    """
    # Standard Model Background (Label 0): Dense, lower energy region
    bg_pt = np.random.normal(loc=25.0, scale=5.0, size=num_events // 2)
    bg_eta = np.random.normal(loc=0.0, scale=1.2, size=num_events // 2)
    bg_data = np.column_stack((bg_pt, bg_eta))
    bg_labels = np.zeros(num_events // 2)

    # Exotic New Physics Anomaly (Label 1): High pT, wide angular spread
    sig_pt = np.random.normal(loc=45.0, scale=8.0, size=num_events // 2)
    sig_eta = np.random.normal(loc=1.8, scale=0.8, size=num_events // 2)
    sig_data = np.column_stack((sig_pt, sig_eta))
    sig_labels = np.ones(num_events // 2)

    X = np.vstack((bg_data, sig_data))
    y = np.hstack((bg_labels, sig_labels))

    return X, y

# Generate and scale dataset
X_raw, y = generate_lhc_event_data(num_events=160)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# Split into Train and Test sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Dataset Ready: {len(X_train)} Training Events, {len(X_test)} Testing Events")
from qiskit.circuit.library import ZZFeatureMap
from qiskit_machine_learning.kernels import FidelityQuantumKernel

# Define 2-Qubit Feature Map (1 Qubit per particle collision feature)
feature_dim = 2
feature_map = ZZFeatureMap(feature_dimension=feature_dim, reps=2, entanglement='linear')

# Construct Quantum Kernel based on Statevector fidelity
quantum_kernel = FidelityQuantumKernel(feature_map=feature_map)

# Print Quantum Circuit Diagram
print("--- Quantum Feature Map Circuit ---")
print(feature_map.decompose().draw(output='text'))
from qiskit_machine_learning.algorithms import QSVC
from sklearn.metrics import accuracy_score, classification_report

# Initialize and Train QSVC using the Quantum Kernel
qsvc = QSVC(quantum_kernel=quantum_kernel)
print("Training Quantum Classifier on Simulated Quantum Processor...")
qsvc.fit(X_train, y_train)

# Predict on Test Events
y_pred = qsvc.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("\n" + "="*45)
print(f" Quantum Model Classification Accuracy: {accuracy * 100:.2f}%")
print("="*45)
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=['SM Background', 'Exotic Anomaly']))
# Create mesh grid for decision boundary visualization
x_min, x_max = X_scaled[:, 0].min() - 0.5, X_scaled[:, 0].max() + 0.5
y_min, y_max = X_scaled[:, 1].min() - 0.5, X_scaled[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 30), np.linspace(y_min, y_max, 30))

grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = qsvc.predict(grid_points)
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6), dpi=140)
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X_test[y_test == 0, 0], X_test[y_test == 0, 1], color='blue', label='SM Background', edgecolors='k', s=50)
plt.scatter(X_test[y_test == 1, 0], X_test[y_test == 1, 1], color='red', label='Exotic Anomaly', edgecolors='k', s=50)

plt.title("Quantum Support Vector Classifier (QSVC)\nLHC Particle Collision Anomaly Detection", fontsize=11, fontweight='bold')
plt.xlabel("Scaled Transverse Momentum ($p_T$)", fontsize=10)
plt.ylabel("Scaled Pseudorapidity ($\eta$)", fontsize=10)
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
